```python
import pandas as pd
import os
from pathlib import Path

# ==================== 请修改这里的路径 ====================
# 原始数据文件路径（请修改为您的实际路径）
input_file = r"C:\Users\Dell\Desktop\QC外廠收貨記錄.xlsx"

# 输出文件路径（桌面）
output_file = os.path.join(os.path.expanduser("~"), "Desktop", "output.xlsx")
# ========================================================

def main():
    print("正在读取数据...")
    # 读取 Excel 文件
    df = pd.read_excel(input_file, sheet_name="查詢1")
    
    # 重命名列（方便处理）
    df.columns = ["Name", "日期", "供应商", "款号", "批号", "数量", "抽查数量", "结果", "抽查结果", "备注", "res_qty"]
    
    # 确保数量是数值类型
    df["数量"] = pd.to_numeric(df["数量"], errors="coerce")
    
    # 过滤掉数量为空的数据
    df = df.dropna(subset=["数量"])
    
    # 确保日期是 datetime 类型
    df["日期"] = pd.to_datetime(df["日期"])
    
    # 提取年份、月份、周、日期
    df["年"] = df["日期"].dt.year
    df["月"] = df["日期"].dt.month
    df["日"] = df["日期"].dt.day
    df["周"] = df["日期"].dt.isocalendar().week
    
    # 创建用于存储结果的列表
    results = []
    
    # 1. 按供应商 + 年统计
    print("正在统计 供应商 + 年...")
    for (supplier, year), group in df.groupby(["供应商", "年"]):
        total_qty = group["数量"].sum()
        avg_qty = group["数量"].mean()
        count = len(group)
        results.append({
            "维度": "年",
            "供应商": supplier,
            "年份": year,
            "月份": None,
            "周": None,
            "日期": None,
            "总数量": total_qty,
            "平均数量": round(avg_qty, 1),
            "收貨筆數": count
        })
    
    # 2. 按供应商 + 年 + 月统计
    print("正在统计 供应商 + 年 + 月...")
    for (supplier, year, month), group in df.groupby(["供应商", "年", "月"]):
        total_qty = group["数量"].sum()
        avg_qty = group["数量"].mean()
        count = len(group)
        results.append({
            "维度": "月",
            "供应商": supplier,
            "年份": year,
            "月份": month,
            "周": None,
            "日期": None,
            "总数量": total_qty,
            "平均数量": round(avg_qty, 1),
            "收貨筆數": count
        })
    
    # 3. 按供应商 + 年 + 周统计
    print("正在统计 供应商 + 年 + 周...")
    for (supplier, year, week), group in df.groupby(["供应商", "年", "周"]):
        total_qty = group["数量"].sum()
        avg_qty = group["数量"].mean()
        count = len(group)
        results.append({
            "维度": "周",
            "供应商": supplier,
            "年份": year,
            "月份": None,
            "周": week,
            "日期": None,
            "总数量": total_qty,
            "平均数量": round(avg_qty, 1),
            "收貨筆數": count
        })
    
    # 4. 按供应商 + 具体日期统计
    print("正在统计 供应商 + 日期...")
    for (supplier, date), group in df.groupby(["供应商", "日期"]):
        total_qty = group["数量"].sum()
        avg_qty = group["数量"].mean()
        count = len(group)
        results.append({
            "维度": "日",
            "供应商": supplier,
            "年份": date.year,
            "月份": date.month,
            "周": None,
            "日期": date.strftime("%Y-%m-%d"),
            "总数量": total_qty,
            "平均数量": round(avg_qty, 1),
            "收貨筆數": count
        })
    
    # 转换为 DataFrame
    result_df = pd.DataFrame(results)
    
    # 排序
    result_df = result_df.sort_values(["供应商", "年份", "月份", "周", "日期"], 
                                       ascending=[True, False, False, False, False])
    
    # 写入 Excel（分不同的 sheet）
    print("正在写入 Excel...")
    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        # 年统计
        year_df = result_df[result_df["维度"] == "年"].drop(columns=["月份", "周", "日期"])
        year_df.to_excel(writer, sheet_name="年統計", index=False)
        
        # 月统计
        month_df = result_df[result_df["维度"] == "月"].drop(columns=["周", "日期"])
        month_df.to_excel(writer, sheet_name="月統計", index=False)
        
        # 周统计
        week_df = result_df[result_df["维度"] == "周"].drop(columns=["月份", "日期"])
        week_df.to_excel(writer, sheet_name="週統計", index=False)
        
        # 日统计
        day_df = result_df[result_df["维度"] == "日"].drop(columns=["月份", "周"])
        day_df.to_excel(writer, sheet_name="日統計", index=False)
        
        # 汇总表（所有供应商的总统计）【已修正】
        summary = result_df.groupby(["供应商"])[["总数量", "收貨筆數"]].sum().reset_index()
        summary["平均数量"] = round(summary["总数量"] / summary["收貨筆數"], 1)
        summary = summary.sort_values("总数量", ascending=False)
        summary.to_excel(writer, sheet_name="供應商總匯總", index=False)
    
    print(f"\n✅ 成功！报表已保存到：{output_file}")
    print("\n📊 生成的 Sheet 包含：")
    print("  - 年統計")
    print("  - 月統計")
    print("  - 週統計")
    print("  - 日統計")
    print("  - 供應商總匯總")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import os
import tempfile
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Agg')
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'Arial']
plt.rcParams['axes.unicode_minus'] = False

# ==================== 请修改这里的路径 ====================
input_file = r"C:\Users\Dell\Desktop\QC外廠收貨記錄.xlsx"
output_file = os.path.join(os.path.expanduser("~"), "Desktop", "output_trend.xlsx")
# ========================================================

def create_trend_charts(df, temp_dir):
    """建立趨勢圖表並儲存為圖片"""
    
    # 確保日期格式
    df["日期"] = pd.to_datetime(df["日期"])
    df["年"] = df["日期"].dt.year
    df["月"] = df["日期"].dt.month
    df["年-月"] = df["日期"].dt.strftime("%Y-%m")
    df["周"] = df["日期"].dt.isocalendar().week
    df["年-周"] = df["年"].astype(str) + "-W" + df["周"].astype(str)
    
    # 過濾掉數量為空的值
    df = df.dropna(subset=["數量"])
    
    # 獲取主要供應商（前5大）
    top_suppliers = df.groupby("供应商")["數量"].sum().nlargest(5).index.tolist()
    
    charts = []
    
    # ========== 圖表1：每月總收貨量趨勢（所有供應商加總）==========
    fig, ax = plt.subplots(figsize=(12, 6))
    monthly_total = df.groupby("年-月")["數量"].sum().sort_index()
    ax.plot(monthly_total.index, monthly_total.values, marker='o', linewidth=2, markersize=6, color='#2E86AB')
    ax.set_title("📈 每月總收貨量趨勢（所有供應商）", fontsize=14, fontweight='bold')
    ax.set_xlabel("月份", fontsize=11)
    ax.set_ylabel("總收貨數量", fontsize=11)
    ax.grid(True, alpha=0.3)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    chart1_path = os.path.join(temp_dir, "temp_chart1.png")
    plt.savefig(chart1_path, dpi=150, bbox_inches='tight')
    plt.close()
    charts.append(("每月總收貨量趨勢", chart1_path))
    
    # ========== 圖表2：前5大供應商每月收貨量變化（分開顯示）==========
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    axes = axes.flatten()
    
    for idx, supplier in enumerate(top_suppliers):
        if idx >= 5:
            break
        supplier_data = df[df["供应商"] == supplier]
        monthly = supplier_data.groupby("年-月")["數量"].sum().sort_index()
        ax = axes[idx]
        ax.bar(monthly.index, monthly.values, color='#A23B72', alpha=0.7)
        ax.set_title(f"{supplier}", fontsize=12, fontweight='bold')
        ax.set_xlabel("月份")
        ax.set_ylabel("收貨數量")
        ax.tick_params(axis='x', rotation=45)
        ax.grid(True, alpha=0.3)
    
    # 隱藏多餘的子圖
    for idx in range(len(top_suppliers), 6):
        axes[idx].set_visible(False)
    
    fig.suptitle("📊 前5大供應商每月收貨量變化", fontsize=14, fontweight='bold')
    plt.tight_layout()
    chart2_path = os.path.join(temp_dir, "temp_chart2.png")
    plt.savefig(chart2_path, dpi=150, bbox_inches='tight')
    plt.close()
    charts.append(("前5大供應商每月收貨量變化", chart2_path))
    
    # ========== 圖表3：每月平均單批數量趨勢（品質穩定性）==========
    fig, ax = plt.subplots(figsize=(12, 6))
    monthly_avg = df.groupby("年-月")["數量"].mean().sort_index()
    ax.plot(monthly_avg.index, monthly_avg.values, marker='s', linewidth=2, markersize=6, color='#F18F01')
    ax.fill_between(monthly_avg.index, monthly_avg.values, alpha=0.3, color='#F18F01')
    ax.set_title("📉 每月平均單批收貨量趨勢（反映批次規模變化）", fontsize=14, fontweight='bold')
    ax.set_xlabel("月份", fontsize=11)
    ax.set_ylabel("平均單批數量", fontsize=11)
    ax.grid(True, alpha=0.3)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    chart3_path = os.path.join(temp_dir, "temp_chart3.png")
    plt.savefig(chart3_path, dpi=150, bbox_inches='tight')
    plt.close()
    charts.append(("每月平均單批收貨量趨勢", chart3_path))
    
    # ========== 圖表4：各供應商每月收貨量堆疊圖 ==========
    fig, ax = plt.subplots(figsize=(14, 7))
    pivot = df.pivot_table(index="年-月", columns="供应商", values="數量", aggfunc="sum", fill_value=0)
    pivot = pivot.sort_index()
    # 只顯示主要供應商
    pivot_top = pivot[top_suppliers]
    pivot_top.plot(kind='area', stacked=True, ax=ax, alpha=0.7)
    ax.set_title("📊 各供應商每月收貨量佔比變化（堆疊圖）", fontsize=14, fontweight='bold')
    ax.set_xlabel("月份", fontsize=11)
    ax.set_ylabel("收貨數量", fontsize=11)
    ax.legend(loc='upper left', bbox_to_anchor=(1, 1))
    ax.grid(True, alpha=0.3)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    chart4_path = os.path.join(temp_dir, "temp_chart4.png")
    plt.savefig(chart4_path, dpi=150, bbox_inches='tight')
    plt.close()
    charts.append(("各供應商每月收貨量佔比變化", chart4_path))
    
    return charts, top_suppliers

def main():
    print("正在讀取數據...")
    
    # 檢查輸入檔案是否存在
    if not os.path.exists(input_file):
        print(f"❌ 錯誤：找不到檔案 {input_file}")
        print("請確認檔案路徑是否正確")
        return
    
    df = pd.read_excel(input_file, sheet_name="查詢1")
    
    # 重命名列
    df.columns = ["Name", "日期", "供应商", "款号", "批号", "數量", "抽查數量", "結果", "抽查結果", "備註", "res_qty"]
    
    # 確保數量是數值類型
    df["數量"] = pd.to_numeric(df["數量"], errors="coerce")
    df = df.dropna(subset=["數量"])
    df["日期"] = pd.to_datetime(df["日期"])
    
    # 提取時間欄位
    df["年"] = df["日期"].dt.year
    df["月"] = df["日期"].dt.month
    df["年-月"] = df["日期"].dt.strftime("%Y-%m")
    df["周"] = df["日期"].dt.isocalendar().week
    df["年-周"] = df["年"].astype(str) + "-W" + df["周"].astype(str)
    
    # 創建暫存目錄
    temp_dir = tempfile.mkdtemp()
    print(f"暫存目錄：{temp_dir}")
    
    print("正在建立趨勢圖表...")
    charts, top_suppliers = create_trend_charts(df, temp_dir)
    
    # ========== 建立 Excel 報表 ==========
    print("正在建立 Excel 報表...")
    
    # 先建立 Excel 檔案
    with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
        
        # 1. 每月趨勢總表
        monthly_summary = df.groupby(["年", "月", "年-月"]).agg(
            總收貨數量=("數量", "sum"),
            平均單批數量=("數量", "mean"),
            收貨批數=("數量", "count")
        ).reset_index()
        monthly_summary = monthly_summary.sort_values(["年", "月"])
        monthly_summary.to_excel(writer, sheet_name="每月趨勢總表", index=False)
        
        # 2. 每週趨勢總表
        weekly_summary = df.groupby(["年", "周", "年-周"]).agg(
            總收貨數量=("數量", "sum"),
            平均單批數量=("數量", "mean"),
            收貨批數=("數量", "count")
        ).reset_index()
        weekly_summary = weekly_summary.sort_values(["年", "周"])
        weekly_summary.to_excel(writer, sheet_name="每週趨勢總表", index=False)
        
        # 3. 前5大供應商每月趨勢
        for supplier in top_suppliers:
            supplier_data = df[df["供应商"] == supplier]
            supplier_monthly = supplier_data.groupby(["年", "月", "年-月"]).agg(
                總收貨數量=("數量", "sum"),
                平均單批數量=("數量", "mean"),
                收貨批數=("數量", "count")
            ).reset_index()
            supplier_monthly = supplier_monthly.sort_values(["年", "月"])
            sheet_name = f"{supplier}_每月趨勢"[:31]
            supplier_monthly.to_excel(writer, sheet_name=sheet_name, index=False)
        
        # 4. 所有供應商每月交叉表
        pivot_monthly = df.pivot_table(
            index="年-月", 
            columns="供应商", 
            values="數量", 
            aggfunc="sum", 
            fill_value=0
        ).sort_index()
        pivot_monthly.to_excel(writer, sheet_name="供應商每月交叉表")
        
        # 5. 趨勢變化分析（月增減率）
        monthly_trend = monthly_summary.copy()
        monthly_trend["月增減率"] = monthly_trend["總收貨數量"].pct_change() * 100
        monthly_trend["月增減率"] = monthly_trend["月增減率"].round(2)
        monthly_trend.to_excel(writer, sheet_name="月增減率分析", index=False)
    
    # ========== 將圖表插入 Excel ==========
    print("正在插入圖表到 Excel...")
    from openpyxl import load_workbook
    from openpyxl.drawing.image import Image as XLImage
    from openpyxl.utils import get_column_letter
    
    wb = load_workbook(output_file)
    
    for sheet_name, chart_path in charts:
        # 確認圖片檔案存在
        if not os.path.exists(chart_path):
            print(f"警告：圖片檔案不存在 {chart_path}")
            continue
            
        if sheet_name not in wb.sheetnames:
            ws = wb.create_sheet(sheet_name)
        else:
            ws = wb[sheet_name]
        
        # 插入圖片
        img = XLImage(chart_path)
        img.width = 800
        img.height = 500
        ws.add_image(img, "A1")
        
        # 調整欄寬
        ws.column_dimensions[get_column_letter(1)].width = 100
    
    # 儲存 Excel
    wb.save(output_file)
    
    # 清理暫存圖片
    print("正在清理暫存檔案...")
    for _, chart_path in charts:
        try:
            if os.path.exists(chart_path):
                os.remove(chart_path)
        except:
            pass
    
    try:
        os.rmdir(temp_dir)
    except:
        pass
    
    print(f"\n✅ 成功！報表已保存到：{output_file}")
    print("\n📊 生成的內容包含：")
    print("  📈 每月趨勢總表")
    print("  📈 每週趨勢總表")
    print("  📈 前5大供應商每月趨勢（個別工作表）")
    print("  📈 供應商每月交叉表")
    print("  📈 月增減率分析")
    print("  📊 4張趨勢圖表（已嵌入 Excel）")
    print(f"\n🔍 前5大供應商：{', '.join(top_suppliers)}")

if __name__ == "__main__":
    main()
```
